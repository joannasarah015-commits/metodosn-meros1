#METODO DE NEWTON-RAPHSON

Consiste en generar una sucesión $ {P_{n}} $ de aproximaciones a la raiz p de la ecuacion $ f(x)=0 $, comenzando con una aproximacion inicial $ P_{0} $ y usando la ecuacion iterativa:

$ P_{n} = P_{n-1} - f(p_{n-1}) / f´(p_{n-1}) $

Veamos un ejemplo:
ejemplo 2.3 $ f(x)=cos x-x=0 $ Usemos el metodo de newton para resolverla



* Función: $f(x) = \cos(x) - x$
* Derivada: $f'(x) = -\sin(x) - 1$
* Fórmula de iteración:
$p_n = p_{n-1} - \frac{\cos(p_{n-1}) - p_{n-1}}{-\sin(p_{n-1}) - 1}$

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def f(x):
  return np.cos(x) - x
```

```python
def df(x):
  return -np.sin(x) - 1
```

```python
def newton(p0, tol):
    error = 100
    Lerror = []
    Lp_i = []
    while error > tol:
        p1 = p0 - f(p0) / df(p0)
        error = np.abs(p1 - p0)
        Lerror.append(error)
        Lp_i.append(p1)
        p0 = p1
    return Lp_i, Lerror
```

```python
p = 0.739085
```

```python

x = np.linspace(0, 1.5, 500)
y = f(x)

plt.figure(figsize=(8, 5))

#Dibuja la función f(x)
plt.plot(x,y,label=r'$f(x) = \cos(x) - x$',color='blue',linewidth=2)
plt.axhline(0,color='black',linestyle='--',linewidth=0.8)
plt.axvline(0,color='black',linestyle='--',linewidth=0.8)
plt.plot(p,f(p),'ro',markersize=8,label=f'Raíz $p \\approx {p:.4f}$') #Marca el punto de la raíz
plt.title('Método de Newton-Raphson: $f(x) = \cos(x) - x$', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)
plt.show()
```

```python
import pandas as pd

tol = 0.000001
error = 100
p0 = 0.5
Lerror = []
Lp_i = []

while error > tol:
    p1 = p0 - (np.cos(p0) - p0) / (-np.sin(p0) - 1)
    error = np.abs(p1 - p0)
    Lerror.append(error)
    Lp_i.append(p1)
    p0 = p1

p = p1

df = pd.DataFrame({
    'Pn': Lp_i,
    'Error': Lerror
})
df.columns = ['Pn', 'Error']
df
```
